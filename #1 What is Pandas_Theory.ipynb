## What is Pandas_Theory

* **`pd.read_excel(r'C:\Users\dell\Desktop\img\MOCK_DATA.xlsx')`** – `pandas` kitabxanası ilə `MOCK_DATA.xlsx` adlı Excel faylını oxuyur və DataFrame şəklində qaytarır.
 - `r` prefiksi yolun (path) içindəki `\` simvollarının xüsusi simvol kimi işlənməsinin qarşısını alır (`raw string`).

---

* **`pd.__version__`** – `pandas` kitabxanasının hazırda istifadə olunan versiyasını göstərir.
- Məsələn: `2.2.2` və ya `1.5.3` kimi nəticə qaytarır.
