## Series Data Structure, isnull, isna_Theory

* **`pd.Series([1, "Ali", 3.5, "Hey"])`** – Fərqli tiplərdən ibarət elementlərlə (`int`, `str`, `float`) bir `pandas.Series` obyekti yaradır.
- `Series` — bir ölçülü məlumat strukturu olub, həm ədədi, həm də mətn tipli dəyərləri saxlaya bilir.



Nəticə təxmini belə olacaq:

```
0       1
1     Ali
2     3.5
3     Hey
dtype: object
```

---


### `pd.Series()` funksiyası — **parametrlərin izahı**

#### pandas.Series(data=None, index=None, dtype=None, name=None, copy=None, fastpath=no_default)


- **`data`**  –  Series-də saxlanacaq məlumatlar. List, array, dict və s. ola bilər.                                         
- **`index`** –  İstəyə bağlı. Series-in indeks (etiket) siyahısı. Əgər verilməzsə, 0-dan başlayan ədədi indeks yaranır.     
- **`dtype`** –  İstəyə bağlı. Məlumatların tipi (məsələn, `int`, `float`, `str`). Əgər verilərsə, data həmin tipə çevrilir. 
- **`name`**  –  İstəyə bağlı. Series-in adı, adətən sütun adı kimi istifadə olunur.                                         
- **`copy`**  –  Boolean (doğru/yalnış). Əgər `True` olarsa, `data`-nın surəti yaradılır, yoxsa referans alınır.             
- **`fastpath`** – Daxili istifadə üçündür, adətən istifadəçiyə lazım deyil.                                                   
---


* **`obj.values`** – `obj` adlı `Series` və ya `DataFrame` obyektindəki bütün dəyərləri `NumPy` massiv formatında qaytarır.

- `obj` – `pandas` obyektidir (`Series` və ya `DataFrame`),
- `values` – obyektin içindəki xam dəyərlərdir (`ndarray` tipində).

---

* **`obj.index`** – `obj` adlı `Series` və ya `DataFrame` obyektinin indekslərini (sətir etiketlərini) qaytarır.

- `obj` – `pandas` obyektidir (`Series` və ya `DataFrame`),
- `index` – obyektin sətrini təmsil edən etiketlər toplusudur (`Index` və ya `RangeIndex` tipində).

---

* **`obj.loc`** – `obj` adlı `Series` və ya `DataFrame` obyektində etiketlərə əsasən seçim etmək üçün istifadə olunan indeksləmə atributudur.

- `obj` – `pandas` obyektidir (`Series` və ya `DataFrame`),
- `loc` – etiket əsaslı sıra və sütun seçimini təmin edir,
- sintaksis: `obj.loc[etiket]` və ya `obj.loc[etiket_sırası, etiket_sütunu]`.

---

* **`ser.astype('int64')`** – `ser` adlı `pandas.Series` obyektinin elementlərini 64-bit tam ədədə (`int64`) çevrir və yeni `Series` qaytarır.

- `ser` – `pandas.Series` obyektidir,
- `astype('int64')` – tip çevirmə funksiyasıdır,
- nəticə yeni `Series` olur, orijinal obyekt dəyişmir.

---

* **`ser.isnull()`** – `ser` adlı `pandas.Series` obyektində hər bir elementin `NaN` (yox, boş) olub-olmadığını göstərən Boolean (`True`/`False`) dəyərdən ibarət yeni `Series` qaytarır.

- `ser` – `pandas.Series` obyektidir,
- `isnull()` – boş dəyərləri tapmaq üçün istifadə olunur,
- nəticə eyni ölçüdə, `True` olan yerlərdə `NaN` dəyər mövcuddur.

---

* **`ser.isna()`** – `ser` adlı `pandas.Series` obyektində hər bir elementin boş (`NaN`) olub-olmadığını yoxlayır və nəticəni Boolean (`True`/`False`) tipində yeni `Series` olaraq qaytarır.

- `ser` – `pandas.Series` obyektidir,
- `isna()` – `NaN` dəyərləri tapmaq üçün istifadə olunur (funksiya `isnull()` ilə eynidir).
