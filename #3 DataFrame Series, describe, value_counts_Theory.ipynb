## DataFrame Series, describe, value_counts_Theory

---

* **`games = pd.read_csv(r'C:\Users\dell\Desktop\img\vgsalesGlobale.csv')`** – göstərilən yoldan CSV faylını oxuyur və `games` adlı DataFrame yaradır.

- **`games.dtypes`** – `games` DataFrame-dəki hər sütunun məlumat tipini (`dtype`) göstərən `Series` qaytarır.

---

* **`games.describe()`** – `games` adlı `pandas.DataFrame`-in ədədi sütunları üzrə əsas statistik göstəriciləri (say, orta, standart sapma, minimum, kvartillər, maksimum) hesablayır və cədvəl şəklində qaytarır.

- Statistik göstəricilər əsasən ədədi verilənlər üçün tətbiq olunur,
- Lazım olsa, **`include='all'`** parametri ilə bütün sütunlar üçün icmal da almaq mümkündür.

---

* **`games.Genre.value_counts()`** – `games` DataFrame-də `Genre` sütunundakı unikal janrların sayını və hər bir janrın neçə dəfə təkrarlandığını göstərir.

- Nəticə sıralanmış formada (`ən çoxdan ən aza` doğru) unikal janrların tezlikləri olur,
- Analiz və vizualizasiya üçün faydalıdır.

---

* **`games.Genre.value_counts(normalize=True)`** – `games` DataFrame-də `Genre` sütunundakı unikal janrların nisbətlərini (faiz və ya proporsiya) göstərir.

- **`normalize=True`** parametri unikal dəyərlərin ümumi say içində faizini hesablayır,
- Nəticə hər janrın ümumi daxilolmalar içindəki payını verir.

---

* **`games.Genre.unique()`** – `games` DataFrame-də `Genre` sütunundakı unikal (təkrarlanmayan) janrları bir `NumPy` massivində qaytarır.

- Sıranı saxlayır,
- Janrların çeşidini görmək üçün istifadə olunur.

---

* **`pd.crosstab(games.Genre, games.Year)`** – `games` DataFrame-də `Genre` və `Year` sütunlarının kəsişməsində tezliklər cədvəli (pivot table) yaradır.

- Satırlarda `Genre`, sütunlarda `Year` yerləşir,
- Hər hüceyrədə həmin janr və ilin kəsişməsində oyunların sayı göstərilir,
- Kateqoriyalar üzrə paylanmanı analiz etmək üçün faydalıdır.

---

* **`games.Global_Sales.mean()`** – `games` DataFrame-də `Global_Sales` sütunundakı satışların orta (ortalama) dəyərini hesablayır.

- Satışların ümumi orta göstəricisini verir,
- Kəmiyyət analizi üçün istifadə olunur.

---

* **`games.Genre.value_counts().plot(kind='bar')`** – `games` DataFrame-də `Genre` sütunundakı janrların sayını sütun qrafiki (bar chart) şəklində vizuallaşdırır.

- `value_counts()` janrların tezliyini hesablayır,
- `plot(kind='bar')` isə bu məlumatı bar chart kimi göstərir,
- Janrların paylanmasını vizual olaraq görmək üçün əlverişlidir.
