## DataFrame, head, tail, loc, iloc, del, T, rename index & columns_Theory



* **`pd.DataFrame(data=data)`** – Verilmiş `data` obyektindən yeni bir `pandas.DataFrame` yaradır.

- `data` – siyahı, lüğət, `NumPy` massiv və ya başqa DataFrame ola bilər,
- `pd.DataFrame()` – `pandas`-ın əsas cədvəl tipli məlumat strukturu yaratma metodudur.

---

### `pd.DataFrame()` funksiyası — **parametrlərin izahı**

#### pandas.DataFrame(data=None, index=None, columns=None, dtype=None, copy=False)

* **`data`** – DataFrame-də saxlanacaq məlumatlar. Ola bilər:

  * 2D array-lər (numpy.array),
  * dict-lər (sözlük) — məsələn: `{‘a’: [1, 2], ‘b’: [3, 4]}`,
  * başqa DataFrame obyektləri,
  * Series-lər listi və s.

* **`index`** – İstəyə bağlı. Satır indekslərinin siyahısı. Verilməzsə, 0-dan başlayan ədədi indeks təyin olunur.

* **`columns`** – İstəyə bağlı. Sütun adlarının siyahısı. Əgər verilməzsə, `data`-dakı sütunlar avtomatik adlandırılır.

* **`dtype`** – İstəyə bağlı. DataFrame-in **bütün elementləri** üçün vahid tip (məsələn `float`, `str`). Əgər `data` uyğundursa, həmin tipə çevrilir.

* **`copy`** – Boolean (doğru/yalnış). Əgər `True` verilsə, `data`-nın surəti yaradılır; əks halda referans istifadə olunur.

---

* **`df.head()`** – `df` adlı `pandas.DataFrame` obyektinin ilk 5 sətrini göstərir.

- Susmaya görə 5 sətir qaytarır,
- İstəyə görə `df.head(n)` şəklində `n` qədər ilk sətiri göstərmək mümkündür.

---

* **`df.tail()`** – `df` adlı `pandas.DataFrame` obyektinin son 5 sətrini göstərir.

- Susmaya görə son 5 sətir qaytarır,
- İstəyə görə `df.tail(n)` formasında `n` qədər son sətiri göstərmək mümkündür.

---

* **`pd.DataFrame(games, columns=['Year', 'Rank', 'Genre'])`** – `games` DataFrame-dən yalnız `Year`, `Rank` və `Genre` sütunlarını seçərək yeni bir `DataFrame` yaradır.

- `games` – mövcud `DataFrame`-dir,
- `columns=[...]` – yalnız göstərilən sütunlar daxil edilir,
- Yeni `DataFrame`-də artıq sütunlar olmayacaq.

---

* **`games.columns = ['Year', 'Rank', 'Platf']`** – `games` DataFrame-dəki bütün sütun adlarını sırasıyla `Year`, `Rank` və `Platf` olaraq yenidən adlandırır.

- Sütun sayı ilə yeni adların sayı eyni olmalıdır,
- Əvvəlki sütun adlarını tamamilə əvəz edir.

---

* **`games.loc`** – `games` adlı `pandas.DataFrame` obyektində **etiketlərə əsasən** sətir və sütun seçimi etmək üçün istifadə olunur.

- `loc` – "label-based selection" metodudur,
- Sintaksis:

  * `games.loc[satır_etiketi]` – bir sətir,
  * `games.loc[satır_etiketi, sütun_etiketi]` – müəyyən sətir və sütun,
  * `games.loc[[etiket1, etiket2], ['Column1', 'Column2']]` – bir neçə sətir və sütun seçimi.

---

* **`games.iloc`** – `games` adlı `pandas.DataFrame` obyektində **pozisiyaya əsasən** (rəqəmlə) sətir və sütun seçimi etmək üçün istifadə olunur.

- `iloc` – "integer-location based selection" metodudur,
- Sintaksis:

  * `games.iloc[0]` – ilk sətri qaytarır,
  * `games.iloc[2, 1]` – 3-cü sətrin 2-ci sütunundakı dəyəri qaytarır,
  * `games.iloc[1:4, 0:2]` – 2-dən 4-ə qədər sətrləri və 1-dən 2-yə qədər sütunları seçir.

---

* **`del games['category']`** – `games` adlı `DataFrame`-dən `'category'` adlı sütunu tamamilə silir.

- Bu əməliyyat yerində (in-place) aparılır,
- Silinən sütun bərpa olunmur,
- Əgər sütun mövcud deyilsə, xəta (`KeyError`) qaytarır.

---

* **`scores_df.index.name = 'name'`** – `scores_df` DataFrame-in sətr (index) etiketlərinə `'name'` adını verir.
* **`scores_df.columns.name = 'lesson'`** – `scores_df` DataFrame-in sütun (columns) etiketlərinə `'lesson'` adını verir.

- Bu adlar nəticənin başlığında görünür,
- Vizual oxunaqlılığı artırmaq üçün istifadə olunur.

---

* **`games.T`** – `games` DataFrame-in transpozunu qaytarır; yəni sətirlər sütunlara, sütunlar isə sətirlərə çevrilir.

- Transpoz əməliyyatı DataFrame-in ölçülərini dəyişir,
- Analiz və vizuallaşdırmada bəzən faydalıdır.

---

* **`games.values.shape`** – `games` DataFrame-dəki xam məlumatların (`NumPy` massiv şəklində) ölçülərini `(sətir sayı, sütun sayı)` kimi qaytarır.

- `values` DataFrame-in dəyərlərini `ndarray` kimi verir,
- `shape` isə massiv ölçüsünü göstərir.

---

* **`scores_df.rename(index={'b': 'A'}, inplace=True)`** – `scores_df` DataFrame-də sətir indekslərindən `'b'` olanı `'A'` ilə əvəz edir.

- `index` parametri indeks adlarının dəyişdirilməsini göstərir,
- `inplace=True` əməliyyatı mövcud DataFrame üzərində yerinə yetirir, yenisini qaytarmır.

---

* **`scores_df.rename(columns={'Math': 'History'}, inplace=True)`** – `scores_df` DataFrame-də `Math` adlı sütunun adını `History` ilə əvəz edir.

- `columns` parametri sütun adlarının dəyişdirilməsini göstərir,
- `inplace=True` əməliyyatın mövcud DataFrame-də aparılmasını təmin edir.

---

* **`games.index`** – `games` DataFrame-in indekslərini (sətir etiketlərini) qaytarır.

- Bu, `RangeIndex` və ya xüsusi etiketlər ola bilər,
- İndeks sətirlərin unikal identifikatorudur.

---

* **`games.columns`** – `games` DataFrame-dəki sütun adlarının siyahısını (`Index` obyektini) qaytarır.

- Sütun adları DataFrame-in dəyişənlərini göstərir,
- Məlumat strukturu və analiz üçün vacibdir.
