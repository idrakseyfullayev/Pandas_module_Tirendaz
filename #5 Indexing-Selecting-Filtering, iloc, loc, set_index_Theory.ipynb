## Indexing-Selecting-Filtering, iloc, loc, set_index


* **`games[games["Year"] > 2010]`** – `games` DataFrame-ində yalnız `Year` sütunundakı dəyəri `2010`-dan böyük olan sətrləri filtr edir və həmin sətrlərdən ibarət yeni DataFrame qaytarır.

- Şərt: `Year > 2010`
- Filtr nəticəsi: müasir illərə aid oyun məlumatları

---

* **`games.iloc`** – `games` DataFrame-də **pozisiyaya əsasən** (yəni sətir və sütun nömrələrinə görə) məlumat seçmək üçün istifadə olunur.

- `iloc` — *integer-location based indexing* deməkdir,
- Sintaksis nümunələri:

  * `games.iloc[0]` – ilk sətri verir,
  * `games.iloc[2, 1]` – 3-cü sətrin 2-ci sütunundakı dəyəri verir,
  * `games.iloc[1:4, 0:2]` – 2–4-cü sətrləri və 1–2-ci sütunları qaytarır.

---

* **`games.loc`** – `games` DataFrame-də **etiketlərə əsaslanaraq** (yəni sətir/sütun adları ilə) məlumat seçmək üçün istifadə olunur.

- `loc` — *label-based indexing* deməkdir,
- Sintaksis nümunələri:

  * `games.loc[5]` – indeks etiketi 5 olan sətri verir,
  * `games.loc[5, 'Genre']` – indeks etiketi 5 olan sətrin `Genre` sütunundakı dəyəri,
  * `games.loc[3:7, ['Genre', 'Year']]` – 3-dən 7-yə qədər indeks aralığında `Genre` və `Year` sütunlarını verir.

---

* **`games.set_index('Genre')`** – `games` DataFrame-də `Genre` sütununu indeks (sətir etiketi) kimi təyin edir və nəticədə yeni bir DataFrame qaytarır.

- Əvvəlki indeks əvəz olunur,
- `Genre` sütunu artıq ayrıca sütun kimi göstərilmir,
- Əgər `inplace=True` yazılmasa, orijinal `games` DataFrame dəyişməz qalır.

---

* **`games[::-1]`** – `games` DataFrame-dəki sətrləri tərsinə (əks sırayla) sıralayır.

- `[::-1]` — Python dilində dilimləmə (slicing) sintaksisidir,
- İlk sətr ən son, sonuncu sətr isə ilk olur,
- DataFrame-in indeksləri dəyişmir, yalnız sıralama tərsinə çevrilir.
