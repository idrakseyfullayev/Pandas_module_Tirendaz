## Methods reindex, drop, add row, mean_Theory



* **`df.reindex()`** – `df` adlı `DataFrame` və ya `Series` obyektinin **sətir və ya sütun sırasını yenidən qurmaq** üçün istifadə olunan metoddur.

#### DataFrame.reindex(labels=None, *, index=None, columns=None, axis=None, method=None, copy=None, level=None, fill_value=nan, limit=None, tolerance=None)


- Sintaksis:

  * `df.reindex(new_index)` – sətir indekslərini dəyişir,
  * `df.reindex(columns=new_columns)` – sütun sırasını dəyişir,
  * `df.reindex(index=new_index, columns=new_columns)` – hər ikisini eyni anda dəyişir.

- Əlavə parametrlər:

  * `method='ffill'` – əvvəlki dəyərlə doldur,
  * `method='bfill'` – sonrakı dəyərlə doldur,
  * `fill_value=0` – `NaN` yerlərə müəyyən dəyəri yaz.

---

* **`ser.reindex(index=["c", "d", "a", "b"])`** – `ser` adlı `Series` obyektinin sətir (indeks) sırasını göstərilən formada (`"c"`, `"d"`, `"a"`, `"b"`) yenidən düzür.

- Mövcud indekslərdən olmayanlar (`"d"` varsa-yoxsa) `NaN` ilə əvəz olunur,
- Yeni sıralama ilə `Series` obyektinin surətini qaytarır (orijinal dəyişmir).

---

* **`ser.reindex(index=range(6), method='ffill')`** – `ser` adlı `Series` obyektinin indeksini `0`-dan `5`-ə qədər yenidən qurur və **boş qalan yerləri əvvəlki dəyərlə doldurur**.

- `range(6)` – yeni indeks: `[0, 1, 2, 3, 4, 5]`,
- `method='ffill'` – **"forward fill"**, yəni boş yerləri əvvəlki mövcud dəyərlə doldurur,
- Orijinal `ser` dəyişmir, nəticə yeni `Series` kimi qaytarılır.

---

* **`df.reindex(columns=names)`** – `df` adlı `DataFrame` obyektində sütunların sırasını `names` siyahısına uyğun şəkildə yenidən düzür.

- `names` – yeni sütun adları sırasını ehtiva edən siyahıdır,
- Əgər `names` siyahısında `df`-də olmayan ad varsa, həmin sütunlar `NaN` ilə doldurulur,
- Yeni `DataFrame` qaytarır; orijinal `df` dəyişmir.


---

* **`games.drop(labels, axis=0, index=None, columns=None, inplace=False, errors='raise')`** – `games` adlı DataFrame-dən sətir və ya sütunları göstərilən adlara və ya indekslərə əsasən silir.

- `labels` – silinəcək sətir və ya sütun adları/indeksləri,
- `axis=0` – sətirləri, `axis=1` – sütunları silmək üçün istifadə olunur,
- `inplace=False` – orijinal DataFrame dəyişmir, yeni DataFrame qaytarılır,
- `errors='raise'` – tapılmayan adlar varsa xəta verir, `'ignore'` seçilsə isə xəta vermir.

* `index` – silinməsi istənilən sətir etiketləri,
* `columns` – silinməsi istənilən sütun adları,
* Bu parametrlər `labels` ilə alternativ olaraq və ya əlavə olaraq istifadə edilə bilər

---

* **`ser.drop(labels=None, axis=0, inplace=False, errors='raise')`** – `ser` adlı `Series` obyektindən göstərilən indeks(lər)ə əsasən element(lər)i silir.

- `labels` – silinəcək elementin və ya elementlərin indeks(lər)i,
- `axis=0` – `Series` bir ölçülü olduğu üçün dəyişmir, default olaraq qalır,
- `inplace=False` – orijinal `ser` dəyişmir, yeni `Series` qaytarılır,
- `errors='raise'` – indeks tapılmazsa xəta verir, `'ignore'` olarsa səhvlər yox sayılır.

---

* **`games.drop(index = games.index[0:5], columns=["Name", "Year"])`** – `games` DataFrame-indən:

-  İlk 5 sətri (`games.index[0:5]`), 
- `Name` və `Year` adlı sütunları eyni anda silir və nəticədə yeni bir DataFrame qaytarır.

* `index=games.index[0:5]` – ilk 5 sətrin indekslərini göstərir,
* `columns=["Name", "Year"]` – göstərilən iki sütunu silir,
* `inplace=False` (susmaya görə) – orijinal `games` dəyişmir.

---

- **`games.drop(index=pd.RangeIndex(0, 5), columns=["Name", "Year"])`** – `games` DataFrame-indən:

* `0`-dan `4`-ə qədər olan sətrləri (ilk 5 sətir),
* `"Name"` və `"Year"` adlı sütunları eyni anda silir və nəticədə yeni bir DataFrame qaytarır.

- `index=pd.RangeIndex(0, 5)` – ilk 5 sətrin indekslərini təyin edir,
- `columns=["Name", "Year"]` – bu iki sütunu silir,
- `inplace=False` (susmaya görə) – orijinal `games` dəyişmir, dəyişiklik yeni nüsxədə olur.

---

* **`games.drop(index=games[games["Name"] == 'Yams'].index)`** – `games` DataFrame-ində `Name` sütununda dəyəri `'Yams'` olan sətrləri tapır və həmin sətrləri silir.

- `games["Name"] == 'Yams'` – `Name` sütununda `'Yams'` olan sətrləri seçir,
- `.index` – həmin sətrlərin indekslərini çıxarır,
- `drop(index=...)` – bu indeksli sətrləri silir.

> Bu əməliyyat yeni DataFrame qaytarır, orijinal `games` dəyişməz qalır (əgər `inplace=True` yazılmayıbsa).

---

* **`df.mean(axis=0, skipna=True, numeric_only=False, **kwargs)`** – `df` adlı `DataFrame`-də ədədi dəyərlərin ortalamasını (mean) hesablayır.

- `axis=0` – sütunlar üzrə ortalama hesablayır (default),
- `axis=1` – hər sətrin ortalamasını hesablayır.
- `skipna=True` – `NaN` dəyərləri nəzərə almadan hesablayır,`False` olsa, `NaN` varsa nəticə də `NaN` olar.
- **`numeric_only=False`** –
  * `True`: yalnız ədədi sütunlar üçün ortalama hesablayır,
  * `False`: mümkün olduqda digər sütunlarda da tətbiq etməyə çalışır.
- `**kwargs` – əlavə parametrlər ötürmək üçündür (məsələn, `level` və s.).
