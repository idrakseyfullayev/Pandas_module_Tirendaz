## Arithmetic Operations add, sub, mul, div, abs, apply_Theory



* **`df1.add(df2, level=None, fill_value=None, axis=1)`** – `df1` DataFrame-i ilə `df2` DataFrame-i və ya Series-i **element üzrə toplama** əməliyyatı aparır.

- `df2` – toplama ediləcək ikinci DataFrame və ya Series obyektidir,
- `level=None` – çoxsəviyyəli indekslərlə işləyərkən müəyyən səviyyəyə görə uyğunlaşdırma aparmaq üçün istifadə olunur,
- `fill_value=None` – uyğun gəlməyən və ya `NaN` olan dəyərlər üçün istifadə olunacaq **əvəzedici dəyər**, məsələn, `fill_value=0` seçilərsə, `NaN` əvəzinə `0` götürüləcək,
- `axis=1`– əməliyyatın hansı ox üzrə aparılacağını göstərir:

  * `axis=0` – **sətir indekslərinə görə** uyğunlaşdırılır (default),
  * `axis=1` – **sütun adlarına görə** uyğunlaşdırılır.


---

* **`df1.add(df2, level=None, fill_value=10, axis=0)`** – `df1` və `df2` DataFrame-ləri arasında **element üzrə toplama** aparır, uyğun gəlməyən və ya `NaN` olan dəyərlər **10 ilə əvəz olunur**.

- **`df2`** – toplama ediləcək ikinci DataFrame və ya Series obyektidir,
- **`level=None`** – çoxsəviyyəli indekslərdə uyğunlaşdırma üçün səviyyəni göstərir (istifadə olunmayıb),
- **`fill_value=10`** – `NaN` və ya uyğun gəlməyən dəyərlər üçün **əvəzedici dəyər olaraq 10 istifadə olunur**,
- **`axis=0`** – toplama **sətir indekslərinə görə** aparılır (default).

---

* **`df1.mul(df2, level=None, fill_value=10, axis=0)`** – `df1` və `df2` DataFrame-ləri arasında **element üzrə vurma** aparır, uyğun gəlməyən və ya `NaN` olan dəyərlər **10 ilə əvəz olunur**.

- **`df2`** – vurma ediləcək ikinci DataFrame və ya Series obyektidir,
- **`level=None`** – çoxsəviyyəli indekslərdə uyğunlaşdırma üçün səviyyəni göstərir (istifadə olunmayıb),
- **`fill_value=10`** – `NaN` və ya uyğun gəlməyən dəyərlər üçün **əvəzedici dəyər olaraq 10 istifadə olunur**,
- **`axis=0`** – vurma **sətir indekslərinə görə** aparılır (default).

---

* **`df1.div(df2, level=None, fill_value=10, axis=0)`** – `df1` və `df2` DataFrame-ləri arasında **element üzrə bölmə** aparır, uyğun gəlməyən və ya `NaN` olan dəyərlər **10 ilə əvəz olunur**.

- **`df2`** – bölmə ediləcək ikinci DataFrame və ya Series obyektidir,
- **`level=None`** – çoxsəviyyəli indekslərdə uyğunlaşdırma üçün səviyyəni göstərir (istifadə olunmayıb),
- **`fill_value=10`** – `NaN` və ya uyğun gəlməyən dəyərlər üçün **əvəzedici dəyər olaraq 10 istifadə olunur**,
- **`axis=0`** – bölmə **sətir indekslərinə görə** aparılır (default).

---

* **`df1.sub(df2, level=None, fill_value=10, axis=0)`** – `df1` və `df2` DataFrame-ləri arasında **element üzrə çıxma** aparır, uyğun gəlməyən və ya `NaN` olan dəyərlər **10 ilə əvəz olunur**.

- **`df2`** – çıxma ediləcək ikinci DataFrame və ya Series obyektidir,
- **`level=None`** – çoxsəviyyəli indekslərdə uyğunlaşdırma üçün səviyyəni göstərir (istifadə olunmayıb),
- **`fill_value=10`** – `NaN` və ya uyğun gəlməyən dəyərlər üçün **əvəzedici dəyər olaraq 10 istifadə olunur**,
- **`axis=0`** – çıxma **sətir indekslərinə görə** aparılır (default).

---


* **`df.abs()`** – DataFrame-dəki bütün ədədi dəyərlərin **modulunu (mənfi işarəsiz dəyərini)** hesablayır.

- Mənfi dəyərləri müsbətə çevirir, müsbət və sıfır dəyərləri olduğu kimi saxlayır.
- Ədədi olmayan (məsələn, mətn) sütunlara təsir etmir.
- İstənilən DataFrame və ya Series-də tətbiq oluna bilər.

---

* **`df.apply(func, axis=0, raw=False, result_type=None, args=(), **kwargs)`** – DataFrame-in hər bir sətirinə və ya sütununa **istifadəçinin verdiyi funksiyanı tətbiq edir**.

- **`func`** – tətbiq olunacaq funksiya (məsələn, `np.sum`, öz yazdığın funksiya və s.),
- **`axis=0`** – funksiyanı **hər sütuna görə** tətbiq edir (default),
  **`axis=1`** – funksiyanı **hər sətirə görə** tətbiq edir,
- **`raw=False`** – funksiyaya `Series` verilir (default),
  **`raw=True`** – funksiyaya `ndarray` verilir,
- **`result_type`** – nəticənin tipini müəyyən edir (məsələn, `'expand'`, `'reduce'`, `'broadcast'`),
- **`args` və `kwargs`** – funksiyaya əlavə arqumentlər.
