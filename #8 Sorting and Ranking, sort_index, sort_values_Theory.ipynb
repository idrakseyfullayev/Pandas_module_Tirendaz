## Sorting and Ranking, sort_index, sort_values_Theory

* **`df.sort_index(axis=0, level=None, ascending=True, inplace=False, kind='quicksort', na_position='last', sort_remaining=True, key=None)`** – `df` DataFrame-ində **sətir və ya sütun indekslərinə əsasən sıralama** aparır.

- **`axis=0`** – **sətir indekslərinə görə** sıralayır (default),
  **`axis=1`** – **sütun adlarına görə** sıralayır,
- **`level=None`** – çoxsəviyyəli indekslərdə uyğunlaşdırma üçün səviyyəni göstərir (istifadə olunmayıb),
- **`ascending=True`** – sıralama **artan sırayla** aparılır (`False` olsa, azalan sırayla),
- **`inplace=False`** – `True` olarsa, dəyişiklik **yerində aparılır**, nəticə qaytarılmır,
- **`kind='quicksort'`** – istifadə edilən sıralama alqoritmi (`quicksort`, `mergesort`, `heapsort` və s.),
- **`na_position='last'`** – `NaN` olan indekslər **axırda yerləşdirilir** (`'first'` olsa əvvəldə),
- **`sort_remaining=True`** – `level` istifadə olunarsa, digər səviyyələr də sıralanır,
- **`key=None`** – sıralamadan əvvəl indekslərə tətbiq olunacaq funksiya.

---

* **`df.sort_values(by, axis=0, ascending=True, inplace=False, kind='quicksort', na_position='last', ignore_index=False, key=None)`** – `df` DataFrame-ində **müəyyən bir və ya bir neçə sütunun dəyərlərinə əsasən sıralama** aparır.

- **`by`** – sıralama aparılacaq **sütun adı və ya adlarının siyahısıdır**,
- **`axis=0`** – sıralama **sətirlər üzrə** aparılır (default),
  **`axis=1`** – sıralama **sütunlar üzrə** aparılır (çox nadir istifadə olunur),
- **`ascending=True`** – sıralama **artan sıra ilə** aparılır (`False` olsa, azalan),
- **`inplace=False`** – `True` olarsa dəyişiklik **yerində aparılır**, nəticə qaytarılmır,
- **`kind='quicksort'`** – istifadə edilən sıralama alqoritmi (`quicksort`, `mergesort`, `heapsort` və s.),
- **`na_position='last'`** – `NaN` olan dəyərlər **axırda yerləşdirilir** (`'first'` olsa əvvəldə),
- **`ignore_index=False`** – `True` olsa, nəticədə indeks sıfırdan başlayaraq yenidən verilir,
- **`key=None`** – sıralamadan əvvəl dəyərlərə tətbiq olunacaq funksiya.
