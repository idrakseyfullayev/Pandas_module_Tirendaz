## Descriptive Statistics_Theory

* **`df.sum(axis=0, skipna=True, numeric_only=False, min_count=0, **kwargs)`** – `df` DataFrame-indəki **ədədi dəyərlərin cəmini** hesablayır.

- **`axis=0`** – **sütunlar üzrə cəmləyir** (default),
- **`axis=1`** – **sətirlər üzrə cəmləyir**,
- **`skipna=True`** – `NaN` dəyərlər **nəzərə alınmır**,
- **`numeric_only=False`** – `True` olduqda yalnız **ədədi sütunlar** cəmlənir,
- **`min_count=0`** – nəticə üçün **minimum tələb olunan qeyri-NaN dəyər sayı** (əks halda nəticə `NaN` olur),
- **`**kwargs`** – əlavə arqumentlər (`level` və s.) ötürmək üçün istifadə olunur.

---

* **`df.mean(axis=0, skipna=True, numeric_only=False, **kwargs)`** – `df` DataFrame-indəki **ədədi dəyərlərin orta qiymətini (mean)** hesablayır.

- **`axis=0`** – **sütunlar üzrə orta** hesablanır (default),
- **`axis=1`** – **sətirlər üzrə orta** hesablanır,
- **`skipna=True`** – `NaN` dəyərlər **nəzərə alınmır**,
- **`numeric_only=False`** – `True` olarsa, yalnız **ədədi sütunlar** üçün hesablanır,
- **`**kwargs`** – əlavə arqumentlər (`level` və s.) ötürmək üçün istifadə olunur.

---

* **`df.idxmax(axis=0, skipna=True, numeric_only=False, **kwargs)`** – `df` DataFrame-ində **ən böyük (maksimum) dəyərin indeksini** (sətir və ya sütun adı) qaytarır.

- **`axis=0`** – **hər sütun üzrə maksimum dəyərin olduğu sətir indeksini** qaytarır (default),
- **`axis=1`** – **hər sətir üzrə maksimum dəyərin olduğu sütun adını** qaytarır,
- **`skipna=True`** – `NaN` dəyərlər **nəzərə alınmır**,
- **`numeric_only=False`** – `True` olarsa, yalnız **ədədi dəyərlər** olan sütunlarda axtarış aparılır,
- **`**kwargs`** – əlavə arqumentlər (`level` və s.) ötürmək üçün istifadə olunur.

---

* **`df.idxmin(axis=0, skipna=True, numeric_only=False, **kwargs)`** – `df` DataFrame-ində **ən kiçik (minimum) dəyərin indeksini** (sətir və ya sütun adı) qaytarır.

- **`axis=0`** – **hər sütun üzrə minimum dəyərin olduğu sətir indeksini** qaytarır (default),
- **`axis=1`** – **hər sətir üzrə minimum dəyərin olduğu sütun adını** qaytarır,
- **`skipna=True`** – `NaN` dəyərlər **nəzərə alınmır**,
- **`numeric_only=False`** – `True` olarsa, yalnız **ədədi dəyərlər** olan sütunlarda axtarış aparılır,
- **`**kwargs`** – əlavə arqumentlər (`level` və s.) ötürmək üçün istifadə olunur.

---

* **`df.cumsum(axis=0, skipna=True, numeric_only=False, **kwargs)`** – `df` DataFrame-indəki **ədədi dəyərlərin kümülatif (yığılmış) cəmini** hesablayır.

- **`axis=0`** – **sütunlar üzrə yığılmış cəm** hesablanır (default),
- **`axis=1`** – **sətirlər üzrə yığılmış cəm** hesablanır,
- **`skipna=True`** – `NaN` dəyərlər **nəzərə alınmır**, arada `NaN` olsa, nəticə `NaN` olaraq davam edir,
- **`numeric_only=False`** – `True` olarsa, yalnız **ədədi sütunlar** üçün tətbiq olunur,
- **`**kwargs`** – əlavə arqumentlər (`level` və s.) ötürmək üçün istifadə olunur.

---

* **`df.describe(percentiles=None, include=None, exclude=None, datetime_is_numeric=False)`** – `df` DataFrame-dəki **ədədi və ya kateqorik sütunların statistik xülasəsini** çıxarır.

- **`percentiles=None`** – göstəriləcək **faiz göstəriciləri** (məsələn, `[0.25, 0.5, 0.75]`),
- **`include=None`** – daxil ediləcək **sütun tipləri** (`'all'`, `['object']`, `['number']`, və s.),
- **`exclude=None`** – **istifadə olunmayacaq** sütun tipləri,
- **`datetime_is_numeric=False`** – `True` olarsa, datetime sütunlar **ədədi kimi** analiz olunur.



#### Nəticəyə daxil olan əsas statistikalar (ədədi sütunlar üçün):

* `count` – boş olmayan dəyərlərin sayı
* `mean` – ədədi orta
* `std` – standart sapma
* `min` – minimum dəyər
* `25%`, `50%`, `75%` – kvartillər
* `max` – maksimum dəyər

Kateqorik sütunlar üçün isə:

* `count`, `unique`, `top`, `freq` statistikaları göstərilir.

---

* **`df.corr(method='pearson', min_periods=1)`** – `df` DataFrame-dəki ədədi sütunlar arasında **korrelyasiya matrisini** hesablayır.

- **`method='pearson'`** – korrelyasiya hesablanma üsulu (default). Dəstəklənənlər:

  * `'pearson'` – Pearson korrelyasiya (xətti əlaqə),
  * `'kendall'` – Kendall Tau,
  * `'spearman'` – Spearman rank korrelyasiyası,
- **`min_periods=1`** – korrelyasiya üçün minimum tələb olunan qeyri-NaN dəyərlərin sayı,
- Nəticə olaraq, hər iki sütun arasındakı korrelyasiya dəyərlərindən ibarət **DataFrame (matris)** qaytarılır.

---

* **`df.corrwith(other, axis=0, drop=False, method='pearson')`** – `df` DataFrame-dəki hər bir sütunla `other` (DataFrame və ya Series) arasındakı **korrelyasiya əmsalını** hesablayır.

- **`other`** – korrelyasiya aparılacaq başqa DataFrame və ya Series,
- **`axis=0`** – `df`-də sütunlarla `other`-in uyğun sütunları arasındakı korrelyasiya (default),
  **`axis=1`** – `df`-də sətirlərlə `other`-in uyğun sətirləri arasındakı korrelyasiya,
- **`drop=False`** – `True` olarsa, uyğun gəlməyən indekslər kənar edilir,
- **`method='pearson'`** – istifadə olunan korrelyasiya üsulu (`'pearson'`, `'kendall'`, `'spearman'`),
- Nəticə olaraq, `Series` tipində korrelyasiya dəyərləri qaytarılır.

---

* **`df.cov(min_periods=None, ddof=1, numeric_only=False, **kwargs)`** – `df` DataFrame-dəki ədədi sütunlar arasında **kovariasiya matrisini** hesablayır.

- **`min_periods=None`** – kovariasiya hesablanması üçün **minimum qeyri-NaN müşahidə sayı**,
- **`ddof=1`** – **azadlıq dərəcəsi** (default 1, yəni statistik kovariasiya),
- **`numeric_only=False`** – `True` olarsa, yalnız **ədədi sütunlar** üçün tətbiq olunur,
- **`**kwargs`** – əlavə arqumentlər (`level` və s.) ötürmək üçün istifadə olunur.

> Kovariasiya – iki dəyişənin **birlikdə necə dəyişdiyini** göstərir, amma korrelyasiya kimi **standartlaşdırılmış ölçü** deyil.

> Korrelyasiya əmsalı `[-1, 1]` aralığında olduğu halda, **kovariasiya istənilən ədədi qiyməti** ala bilər.

---

* **`pd.unique(values)`** – Verilmiş `values` massivindəki **unikal (təkrarlanmayan) dəyərləri** qaytarır.

- **`values`** – `Series`, `DataFrame` sütunu və ya `Numpy array` kimi **1 ölçülü** massiv ola bilər,
- Təkrarlanan dəyərlər **yalnız bir dəfə saxlanılır**,
- **Orijinal ardıcıllıq** qorunur, **sıralama aparılmır**,
- Nəticə **NumPy array** formatında qaytarılır.

---

* **`df.value_counts(subset=None, normalize=False, sort=True, ascending=False, dropna=True)`** – `df` DataFrame-dəki **sətirlərin (kombinasiya olaraq) neçə dəfə təkrarlandığını** sayır.

- **`subset=None`** – yalnız müəyyən **sütun(lar)** üzrə təkrarlanan sətirləri saymaq üçün istifadə olunur,
- **`normalize=False`** – `True` olarsa, nəticə **nisbət (%) şəklində** qaytarılır,
- **`sort=True`** – nəticə **saylara görə sıralanır**,
- **`ascending=False`** – sıralama **azalan sıra ilə** aparılır,
- **`dropna=True`** – `NaN` olan sətirlər **sayılmır**, `False` olsa sayılır,
- Nəticə **Series** formatında qaytarılır və **sətirlərin unikal kombinasiyası** indeks kimi göstərilir.

---

* **`df.count(axis=0, level=None, numeric_only=False)`** – `df` DataFrame-dəki **boş olmayan (NaN olmayan) dəyərlərin sayını** hesablayır.

- **`axis=0`** – **hər sütundakı** boş olmayan dəyərlərin sayını verir (default),
- **`axis=1`** – **hər sətirdəki** boş olmayan dəyərlərin sayını verir,
- **`level=None`** – çoxsəviyyəli indekslərdə səviyyəyə görə qruplaşdırılmış sayları qaytarır,
- **`numeric_only=False`** – `True` olduqda yalnız **ədədi dəyərlər** olan sütunlar üçün hesablanır.

> Nəticə `Series` formatında olur: indekslər ya sütun adları, ya da sətir indeksləridir.

---

* **`df.isin(values)`** – `df` DataFrame-dəki hər bir elementin **göstərilən dəyərlərlə uyğun olub-olmadığını** yoxlayır və nəticəni **Boolean (`True`/`False`)** şəklində qaytarır.

- **`values`** – müqayisə üçün istifadə olunan **scalar**, **list**, **dict**, **Series** və ya **DataFrame** ola bilər:

  * `scalar` → bütün elementlərlə müqayisə olunur,
  * `list`/`set`/`tuple` → hər element onlardan biri ilə uyğunlaşırsa `True`,
  * `dict` → açarlar sütun adlarına uyğun gəlməlidir (`{"col1": [1, 2], "col2": [3, 4]}`),
  * `Series`/`DataFrame` → elementlər uyğun indekslər və sütunlara görə müqayisə olunur.

> Nəticə `True` və `False` dəyərlərdən ibarət **DataFrame** formatında olur – orijinal ölçüləri saxlanılır.
