## Reading and Writing Data, read_table, read_csv, to_csv_Theory

* **`pd.read_table(filepath_or_buffer, sep='\t', delimiter=None, header='infer', names=None, index_col=None, usecols=None, dtype=None, engine=None, skiprows=None, nrows=None, na_values=None, parse_dates=False, encoding=None, **kwargs)`** – **Tab və ya digər ayrıcılarla formatlanmış mətn faylını** (`.txt`, `.dat`, və s.) **DataFrame kimi oxumaq** üçün istifadə olunur.

- **`filepath_or_buffer`** – oxunacaq faylın **yolu və ya obyektidir**,
- **`sep='\t'`** – sətrin **hansı simvola görə bölünəcəyini** göstərir (default: tab `\t`),
- **`delimiter=None`** – `sep` ilə eyni işi görür, biri istifadə olunmalıdır,
- **`header='infer'`** – **başlıqların mövcudluğunu** avtomatik təyin edir (`None` olsa, başlıq yoxdur),
- **`names=None`** – başlıq sütunları üçün **əlavə adlar vermək** üçün,
- **`index_col=None`** – hansı sütunun **indeks kimi** təyin ediləcəyini göstərir,
- **`usecols=None`** – yalnız **istədiyin sütunları** oxumaq üçün,
- **`dtype=None`** – sütunların **məlumat tipini** təyin edir,
- **`engine=None`** – istifadə olunacaq **oxuma mühərriki** (`'c'`, `'python'`),
- **`skiprows=None`** – **başdan neçə sətir atılacaq**,
- **`nrows=None`** – **neçə sətir oxunacaq**,
- **`na_values=None`** – hansı dəyərlərin **NaN** kimi qəbul ediləcəyini göstərir,
- **`parse_dates=False`** – `True` olarsa, tarix tipli sütunlar avtomatik tanınır,
- **`encoding=None`** – faylın **kodlaşdırma növü** (məsələn, `'utf-8'`, `'latin1'`),
- **`**kwargs`** – əlavə parametrlər.

> Fayl **tab ilə ayrılmışdırsa**, `sep='\t'` yazmağa ehtiyac yoxdur – bu default olaraq işləyir.
> `.csv` fayllar üçün isə `pd.read_csv()` daha uyğundur.

---

* **`pd.read_csv(filepath_or_buffer, sep=',', delimiter=None, header='infer', names=None, index_col=None, usecols=None, dtype=None, engine=None, skiprows=None, nrows=None, na_values=None, parse_dates=False, encoding=None, **kwargs)`** – **Vergüllə ayrılmış (.csv) faylları** oxuyaraq **DataFrame** yaradır.

- **`filepath_or_buffer`** – oxunacaq faylın **yolu və ya fayl obyektidir**,
- **`sep=','`** – **sətirlərin hansı simvola görə bölünəcəyini** göstərir (default: vergül `,`),
- **`delimiter=None`** – `sep` ilə eyni işi görür, biri istifadə olunmalıdır,
- **`header='infer'`** – başlıqların **ilk sətrdən götürülüb-götürülməməsini** göstərir,
- **`names=None`** – öz sütun adlarını **əlavə şəkildə təyin etmək** üçün,
- **`index_col=None`** – hansı sütunun **indeks kimi** istifadə olunacağını göstərir,
- **`usecols=None`** – yalnız **istədiyin sütunları** oxumaq üçün (`list`, `callable` və s.),
- **`dtype=None`** – sütunların **tipini əl ilə** təyin etmək üçün (`{'col1': int, 'col2': str}`),
- **`engine=None`** – istifadə olunan **oxuma mühərriki** (`'c'` və ya `'python'`),
- **`skiprows=None`** – **başdan neçə sətirin atılacağını** göstərir,
- **`nrows=None`** – neçə sətir oxunacağını göstərir,
- **`na_values=None`** – hansı dəyərlərin **NaN** kimi qəbul olunacağını göstərir,
- **`parse_dates=False`** – tarix tipli sütunları **avtomatik təyin etmək** üçün `True` yazılır,
- **`encoding=None`** – faylın **kodlaşdırmasını** göstərir (`'utf-8'`, `'latin1'`, və s.),
- **`**kwargs`** – digər əlavə parametrlər.

---

> `.csv` faylları üçün **ən çox istifadə olunan və əsas fayl oxuma funksiyasıdır**.
> Əgər ayrıcı fərqli simvoldursa (məsələn, `;`), `sep=';'` yazmaq kifayətdir.

---

* **`df.to_csv(path_or_buf=None, sep=',', index=True, header=True, columns=None, encoding=None, mode='w', compression='infer', line_terminator=None, quoting=None, quotechar='"', na_rep='', float_format=None, date_format=None, **kwargs)`** – `df` DataFrame-ni **CSV formatında fayla və ya mətn kimi çıxarır**.

- **`path_or_buf=None`** – Faylın **yolu** (`'data.csv'`) və ya `None` olduqda **string kimi nəticə qaytarır**,
- **`sep=','`** – CSV faylında **sütunların ayrılması üçün simvol** (default: vergül),
- **`index=True`** – **DataFrame indeksini də çıxarır**, `False` olsa çıxarılmaz,
- **`header=True`** – **sütun adlarını da yazır**,
- **`columns=None`** – Yalnız **seçilmiş sütunları** çıxarmaq üçün,
- **`encoding=None`** – Faylın **kodlaşdırması** (`'utf-8'`, `'utf-8-sig'`, `'latin1'`, və s.),
- **`mode='w'`** – Yazma rejimi: `'w'` → üzərinə yaz, `'a'` → əlavə et,
- **`compression='infer'`** – Fayl **zip, gzip** və s. ilə sıxılacaqsa, növü təyin edir,
- **`line_terminator=None`** – Sətir sonu üçün istifadə olunacaq simvol (`\n`, `\r\n` və s.),
- **`quoting=None`** – Sətirdəki dəyərlərin **təklifə alınma (quote) tərzi**,
- **`quotechar='"'`** – Sətir dəyərləri üçün **quote simvolu**,
- **`na_rep=''`** – `NaN` dəyərlər CSV-də **nə ilə əvəz olunacaq**,
- **`float_format=None`** – Float dəyərlərin **formatlanması** (məsələn, `"%.2f"`),
- **`date_format=None`** – Tarix sütunlarının formatı.

---

* **`df.to_csv(sys.stdout, sep='|')`** – `df` DataFrame-in **CSV formatında məzmununu konsola (terminala)** yazdırır və **sütunlar `|` simvolu ilə ayrılır**.

- **`sys.stdout`** – **çıxış faylı əvəzinə**, məlumat **birbaşa terminala** yazdırılır,
- **`sep='|'`** – CSV-də **sütun ayırıcı** olaraq **`|`** istifadə olunur,
