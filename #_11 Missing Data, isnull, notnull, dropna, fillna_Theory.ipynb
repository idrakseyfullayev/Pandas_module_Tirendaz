## Missing Data, isnull, notnull, dropna, fillna_Theory



* **`df.isnull()`** – `df` adlı DataFrame-dəki **NaN (boş)** dəyərləri müəyyən edir və uyğun hüceyrələrdə `True`, digərlərində `False` olan eyni ölçülü DataFrame qaytarır.

---

* **`df.isna()`** – `df` adlı DataFrame-dəki **NaN (boş)** dəyərləri müəyyən edir və uyğun hüceyrələrdə `True`, digərlərində `False` olan eyni ölçülü DataFrame qaytarır.

---

* **`df.notnull()`** – `df` adlı DataFrame-dəki **NaN olmayan (dolu)** dəyərləri müəyyən edir və uyğun hüceyrələrdə `True`, NaN olanlarda isə `False` olan eyni ölçülü DataFrame qaytarır.

---

* **`df.dropna(axis=0, how='any', thresh=None, subset=None, inplace=False)`** – `df` adlı DataFrame-dən **NaN olan sətir və ya sütunları** silir.

- `axis=0` – sətirləri, 
- `axis=1` – sütunları silmək üçün istifadə olunur,
- `how='any'` – **istənilən bir** NaN varsa silir, 
- `how='all'` – **bütün** dəyərlər NaN-dirsə silir,
- `thresh=None` – saxlanması üçün **minimum NaN olmayan dəyərlərin** sayı,
- `subset=None` – yalnız müəyyən **sütun(lar)** üzrə NaN yoxlanılır,
- `inplace=False` – orijinal DataFrame dəyişmir, yeni DataFrame qaytarılır.

---

* **`df.fillna(value=None, method=None, axis=None, inplace=False, limit=None)`** – `df` adlı DataFrame-dəki **NaN (boş)** dəyərləri göstərilən dəyərlə və ya metodla əvəz edir.

- `value=None` – NaN dəyərlərin əvəz olunacağı **sabit dəyər**, sözlük və ya Series,
- `method=None` – əvəzləmə üsulu: `'ffill'` (öncəki ilə doldur), `'bfill'` (sonrakı ilə doldur),
- `axis=None` – doldurma istiqaməti: `0` (sütunlar üzrə), `1` (sətirlər üzrə),
- `inplace=False` – orijinal DataFrame dəyişmir, yeni DataFrame qaytarılır,
- `limit=None` – **maksimum neçə NaN** dəyərin əvəz olunacağını göstərir.
