## Data Cleaning, duplicate, drop_duplicates, map, replace, rename_Theory

---

* **`df.duplicated(subset=None, keep='first')`** – `df` adlı DataFrame-dəki **təkrarlanan sətirləri** müəyyən edir və yalnız təkrarlananların `True`, digərlərinin `False` olduğu Series qaytarır.

- `subset=None` – yalnız müəyyən **sütun(lar)** üzrə təkrarlanmanı yoxlamaq üçün istifadə olunur,
- `keep='first'` – ilk təkrarı saxlayır, digərlərini `True` qaytarır;
- `keep='last'` – sonuncunu saxlayır, əvvəlkiləri `True`;
- `keep=False` – bütün təkrarları `True` qaytarır.

---

* **`df.drop_duplicates(subset=None, keep='first', inplace=False, ignore_index=False)`** – `df` adlı DataFrame-dən **təkrarlanan sətirləri** silir və yalnız unikal sətirləri saxlayır.

- `subset=None` – yalnız müəyyən **sütun(lar)** üzrə təkrarlanmanı yoxlamaq üçün istifadə olunur,
- `keep='first'` – ilk təkrarı saxlayır, digərlərini silir;
- `keep='last'` – sonuncunu saxlayır, əvvəlkiləri silir;
- `keep=False` – bütün təkrarları silir,
- `inplace=False` – orijinal DataFrame dəyişmir, yeni DataFrame qaytarılır,
- `ignore_index=False` – `True` olarsa, nəticədə indeks sıfırdan başlayaraq yenidən qurulur.

---

* **`df.map(func, na_action=None)`** – `df` Series obyektindəki **hər bir elementi** göstərilən **funksiya, sözlük və ya digər xəritələmə obyektinə** əsasən dəyişir.

> Qeyd: `map()` yalnız **Series** üçün istifadə olunur, `DataFrame` üçün `applymap()` işlədilir.

* `func` – tətbiq olunacaq **funksiya**, **sözlük (dict)** və ya **Series**,
* `na_action=None` – `na_action='ignore'` seçilərsə, **NaN dəyərlərə toxunulmur**, əks halda funksiyaya verilir.


---

* **`df.replace(to_replace=None, value=None, inplace=False, limit=None, regex=False, method='pad')`** – `df` adlı DataFrame-dəki **müəyyən dəyərləri** yeniləri ilə əvəz edir.

- `to_replace` – əvəz olunacaq **dəyər, siyahı, sözlük və ya regex nümunəsi**,
- `value` – əvəzləyəcək yeni **dəyər, siyahı və ya sözlük**,
- `inplace=False` – orijinal DataFrame dəyişmir, yeni DataFrame qaytarılır,
- `limit=None` – maksimum neçə dəyişiklik ediləcəyini göstərir,
- `regex=False` – `True` seçilərsə, `to_replace` **regulyar ifadə** kimi qəbul olunur,
- `method='pad'` – uyğun metod: `pad` (öncəki ilə doldur), `bfill` (sonrakı ilə doldur) yalnız `to_replace` NaN olanda istifadə olunur.

---

* **`df.index`** – `df` adlı DataFrame-in **sətir indekslərini** (etiketlərini) göstərir və onu təmsil edən **Index obyektini** qaytarır.

Bu obyekt vasitəsilə:

* indeksləri **görüntüləmək**,
* **dəyişdirmək** və ya
* **filtrasiya əməliyyatlarında** istifadə etmək mümkündür.

---

* **`df.values`** – `df` adlı DataFrame-dəki **bütün məlumatları** **NumPy massivində (array)** verir.

Bu massivdə:

* **yalnız dəyərlər** olur (sətir/sütun adları daxil deyil),
* nəticə `numpy.ndarray` formatında olur,
* **məlumat tipləri qarışıqdısa**, massiv `object` tipində ola bilər.

---

* **`df.columns`** – `df` adlı DataFrame-in **sütun adlarını** göstərir və onu təmsil edən **Index obyektini** qaytarır.

Bu obyekt vasitəsilə:

* sütun adlarını **görüntüləmək**,
* **dəyişmək** və ya
* **sütunlara aid əməliyyatlarda** istifadə etmək mümkündür.

---

* **`df.rename(mapper=None, index=None, columns=None, axis=None, inplace=False, errors='ignore')`** – `df` adlı DataFrame-dəki **sətir və ya sütun adlarını** dəyişir.

- `mapper` – ümumi ad dəyişiklikləri üçün **sözlük və ya funksiya**,
- `index` – yalnız **sətir adlarını** dəyişmək üçün sözlük və ya funksiya,
- `columns` – yalnız **sütun adlarını** dəyişmək üçün sözlük və ya funksiya,
- `axis` – `0` və ya `'index'` (sətir), `1` və ya `'columns'` (sütun) dəyişmək üçün,
- `inplace=False` – orijinal DataFrame dəyişmir, yeni DataFrame qaytarılır,
- `errors='ignore'` – uyğun ad tapılmazsa, xəta vermir (`'raise'` olsa xəta verir).
