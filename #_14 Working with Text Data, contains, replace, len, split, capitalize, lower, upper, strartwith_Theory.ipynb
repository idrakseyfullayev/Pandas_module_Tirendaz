## Working with Text Data, contains, replace, len, split, capitalize, lower, upper, strartwith_Theory

---

* **`ser.str.split(pat=None, n=-1, expand=False)`** – `ser` adlı **string Series**-dəki hər bir elementi **müəyyən ayırıcıya (`pat`) əsasən bölür**.

- `pat=None` – **ayırıcı (separator)**; `None` olduqda boşluqlara əsasən bölür,
- `n=-1` – maksimum neçə dəfə bölünsün (`-1` – hamısı),
- `expand=False` – `False` olarsa **nəticə siyahı** şəklində qalır,
  `True` olarsa **DataFrame** kimi sütunlara ayrılır.

---

* **`ser.str.replace(pat, repl, n=-1, case=None, flags=0, regex=False)`** – `ser` adlı **string Series**-dəki hər bir elementdə **`pat` ilə uyğun gələn hissəni `repl` ilə əvəz edir**.

- `pat` – əvəz olunacaq **söz, simvol və ya regex nümunəsi**,
- `repl` – **əvəzləyəcək yeni simvol və ya ifadə**,
- `n=-1` – **əvəzlənəcək maksimum say** (`-1` – hamısı),
- `case=None` – `True` olarsa **case-sensitive**, `False` olarsa fərq qoymur,
- `flags=0` – **regex əməliyyatları üçün bayraqlar** (məsələn, `re.IGNORECASE`),
- `regex=False` – `True` olarsa, `pat` **regulyar ifadə** kimi qəbul olunur.

---

* **`ser.str.contains(pat, case=True, flags=0, na=False, regex=True)`** – `ser` adlı **string Series**-dəki hər bir elementin **`pat` ilə uyğun gəlib-gəlmədiyini** yoxlayır və nəticədə `True` / `False` dəyərləri olan Series qaytarır.

- `pat` – **axtarılacaq simvol, söz və ya regex nümunəsi**,
- `case=True` – **böyük-kiçik hərf fərqini** nəzərə alır, `False` olsa fərq qoymur,
- `flags=0` – regex üçün **əlavə bayraqlar** (`re.IGNORECASE` və s.),
- `na=False` – **NaN** olan dəyərlər üçün qaytarılacaq nəticə,
- `regex=True` – `pat` **regulyar ifadə** kimi işlənir; `False` olsa düz mətn kimi axtarılır.
