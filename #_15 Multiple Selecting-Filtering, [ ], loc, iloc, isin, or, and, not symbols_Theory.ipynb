## Multiple Selecting-Filtering, [ ], loc, iloc, isin, or, and, not symbols_Theory

---

* **`df.loc[]`** – `df` adlı DataFrame-də **etiket (ad) əsasında sətir və sütunları seçmək** üçün istifadə olunur. İstifadəçi **etiketlərə görə filtr**, **şərt**, **sətir və sütun adlarına görə seçim** edə bilər.

- `df.loc[row_label]` – yalnız sətir seçimi,
- `df.loc[row_label, column_label]` – həm sətir, həm sütun seçimi,
- `row_label` və `column_label` – **etiketlər, siyahılar, dilimlər (`:`)** və ya **şərtlər** ola bilər.

---

* **`|`, `&`, `~`** – `pandas`-da **şərt ifadələrini birləşdirmək və tərs çevirmək** üçün istifadə olunan **məntiqi operatorlardır**. Bu operatorlar `Series`-lər üzərində **element-wise (element üzrə)** işləyir və nəticədə `True` / `False` dəyərləri olan Series qaytarır.

- `|` – **və ya** (OR) – iki şərtdən **ən az biri doğru** olarsa, nəticə `True`,
- `&` – **və** (AND) – **hər iki şərt doğru** olarsa, nəticə `True`,
- `~` – **deyil** (NOT) – şərtin **əksini (tərsini)** alır, `True` → `False`, `False` → `True`.

> **Qeyd:** Bu operatorlarla şərt yazarkən **hər bir şərt mütləq mötərizə `( )` içində** olmalıdır.

---

* **`film.iloc[]`** – `film` adlı DataFrame-dən **pozisiya (ədədi indeks) əsasında sətir və ya sütunları seçmək** üçün istifadə olunur. `iloc` yalnız **tam ədədlərlə (integer)** işləyir və **etiketləri deyil, indeks nömrələrini** əsas götürür.

- `film.iloc[5]` – 5-ci indeksdəki sətri qaytarır,
- `film.iloc[2:6]` – 2-dən 5-ə qədər (6 daxil deyil) olan sətirləri seçir,
- `film.iloc[:, 0]` – bütün sətirlərdən yalnız **0-cı sütunu** seçir,
- `film.iloc[1, 2]` – 1-ci sətrin 2-ci sütunundakı dəyəri verir,
- `film.iloc[[0, 3, 7], [1, 2]]` – 0, 3, 7-ci sətirlərin 1 və 2-ci sütunlarını seçir.

---

* **`df.isin(values)`** – `df` adlı DataFrame-dəki hər bir dəyərin **`values` daxilində olub-olmadığını** yoxlayır və uyğun gələn yerlərdə `True`, digərlərində `False` olan eyni ölçülü DataFrame qaytarır.

- `values` – **axtarılan dəyərlər**: tək bir dəyər, siyahı, sözlük, Series və ya DataFrame ola bilər.
