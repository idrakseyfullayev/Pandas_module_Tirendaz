## Hierarchical Indexing, unstack, stack, swaplevel, sort_index, sum, groupby, set_index, reset_index_Theory

---

* **`ser.unstack(level=-1, fill_value=None)`** – `ser` adlı **MultiIndex-li Series**-i **DataFrame-ə çevirir**, yəni **bir səviyyəni sütun kimi** genişləndirir.

- `level=-1` – **sütun kimi çıxarılacaq səviyyə** (ən alt səviyyə susmaya görə),
- `fill_value=None` – **boş yerlərə yazılacaq dəyər**, məsələn `0` və ya `'NaN'`.

Bu funksiya çoxsəviyyəli indeks strukturlarını **ikiölçülü DataFrame** şəklinə gətirmək üçün istifadə olunur.

---

* **`df.stack(level=-1, dropna=True)`** – `ser` və ya `DataFrame` obyektindəki **sütunları alt səviyyəli sətir indekslərinə** çevirir, yəni **DataFrame-i uzun (long) formata salır**.

- `level=-1` – **yığılacaq səviyyə** (susmaya görə sonuncu),
- `dropna=True` – `NaN` dəyərlər **sıralamadan çıxarılır**; `False` olsa saxlanılır.

Bu funksiya xüsusilə **DataFrame-ləri Series-ə** çevirmək və **MultiIndex yaratmaq** üçün istifadə olunur.

---

* **`df.swaplevel(i=-2, j=-1, axis=0)`** – `df` adlı **MultiIndex**-li DataFrame və ya Series-də **iki indeks səviyyəsinin yerlərini dəyişir**.

- `i`, `j` – **yerlərini dəyişəcək indeks səviyyələrinin adları və ya mövqeləri** (ədədi və ya etiket),
- `axis=0` – səviyyələr **sətir indeksində** dəyişdirilir (`axis=1` olsa sütun indeksində).

Bu funksiya **çoxsəviyyəli indeksləri idarə etmək** və daha çevik məlumat qruplaşdırma və seçmə əməliyyatları üçün istifadə olunur.

---

* **`df.sort_index(axis=0, level=None, ascending=True, inplace=False, kind='quicksort', na_position='last')`** – `df` adlı DataFrame və ya Series-də indeksə əsasən sıralama aparır.

- `axis=0` – sətir indeksinə görə sıralayır, `axis=1` sütun indeksinə görə sıralayır,
- `level=None` – çoxsəviyyəli indeks varsa, yalnız göstərilən səviyyəyə görə sıralama aparır (default bütün səviyyələr).
- `ascending=True` – artan sıra ilə sıralayır, `False` olarsa azalan sıra ilə sıralayır,
- `inplace=False` – orijinal DataFrame dəyişmir, sıralanmış yeni DataFrame qaytarılır,
- `kind='quicksort'` – istifadə olunan sıralama alqoritmi (quicksort, mergesort, heapsort və s. ola bilər),
- `na_position='last'` – NaN dəyərlərin sıralamada sonuncu yerdə yerləşdirilməsi (`'first'` seçilərsə əvvəldə olur),


---

* **`df.sum(axis=None, skipna=True, numeric_only=None, min_count=0, **kwargs)`** – `df` adlı DataFrame və ya Series-də ədədi dəyərlərin cəmini hesablayır.

- `axis=0` (default) – hər sütunun cəmini hesablayır (sətir istiqamətində),
- `axis=1` – hər sətrin cəmini hesablayır (sütun istiqamətində),
- `skipna=True` – NaN dəyərləri hesaba qatmır,
- `numeric_only=None` – yalnız ədədi sütunları hesablamaq üçün (`True`/`False` ola bilər),
- `min_count=0` – ən azı neçə ədəd dəyər olmalıdır ki, nəticə hesablansın (default 0, yəni heç dəyər olmaya da nəticə 0 olar).

---

* **`df.groupby(by=None, axis=0, level=None, as_index=True, sort=True, group_keys=True, squeeze=False, observed=False, dropna=True)`** – `df` DataFrame-də verilən `by` sütun(lar)ına və ya funksiyaya əsasən qruplaşdırma aparır.

- `by` – qruplaşdırma üçün istifadə olunacaq sütun adı(sı), indeks səviyyəsi və ya funksiyalar,
- `axis=0` – qruplaşdırma sətirlər üzrə aparılır (default),
- `level=None` – çoxsəviyyəli indeksdə müəyyən səviyyəyə görə qruplaşdırma,
- `as_index=True` – qruplaşdırma sütunları nəticədə indeks kimi qalır,
- `sort=True` – qruplar sıralanır,
- `group_keys=True` – qruplaşdırılmış açarlar nəticəyə əlavə olunur,
- `squeeze=False` – nəticə DataFrame və ya Series şəklində olur,
- `observed=False` – kateqoriyalı dəyişənlər üçün bütün mümkün kateqoriyalar nəzərə alınır,
- `dropna=True` – NaN dəyərləri qruplardan çıxarır.

---

* **`df.set_index(keys, drop=True, append=False, inplace=False, verify_integrity=False)`** – `df` DataFrame-də göstərilən sütunu/sütunları indeks kimi təyin edir.

- `keys` – indeks kimi təyin olunacaq sütun adı və ya adlarının siyahısı,
- `drop=True` – indeksə çevrilən sütun(lar) DataFrame-dən silinir,
- `append=False` – mövcud indeksə əlavə etmək əvəzinə onu əvəz edir,
- `inplace=False` – orijinal DataFrame dəyişmir, yeni DataFrame qaytarılır,
- `verify_integrity=False` – yeni indeksin unikal olub-olmadığını yoxlamır (`True` olarsa, təkrarlanan indekslər olduqda xəta verir).

---

* **`df.reset_index(level=None, drop=False, inplace=False, col_level=0, col_fill='')`** – `df` DataFrame-də indeksləri adi sütunlara çevirir və ya sıfırdan yeni indeks yaradır.

- `level=None` – yalnız göstərilən indeks səviyyəsini sıfırlayır (çoxsəviyyəli indeks üçün),
- `drop=False` – indeks dəyərlərini sütun kimi saxlamaq üçün (`True` olarsa, indeks dəyərləri atılır),
- `inplace=False` – orijinal DataFrame dəyişmir, yeni DataFrame qaytarılır,
- `col_level=0` – sütun adları çoxsəviyyəlidirsə, hansı səviyyəyə əlavə ediləcəyini göstərir,
- `col_fill=''` – yeni sütun adı üçün boş səviyyələrə hansı dəyərin yazılacağını təyin edir.
