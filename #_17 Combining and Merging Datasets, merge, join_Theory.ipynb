## Combining and Merging Datasets, merge, join_Theory



* **`pd.merge(left, right, how='inner', on=None, left_on=None, right_on=None, left_index=False, right_index=False, sort=False, suffixes=('_x', '_y'), copy=True, indicator=False, validate=None)`** – İki DataFrame-i (`left` və `right`) bir və ya bir neçə açara əsasən birləşdirir.

- `left`, `right` – birləşdiriləcək iki DataFrame,
- `how='inner'` – birləşdirmə üsulu: `'left'`, `'right'`, `'outer'`, `'inner'` (default: yalnız uyğun gələnlər),
- `on` – hər iki DataFrame-də olan və birləşdiriləcək sütun(lar),
- `left_on`, `right_on` – ayrı-ayrı sütun adları ilə birləşdirmə imkanı,
- `left_index=False`, `right_index=False` – indekslərə əsasən birləşdirmə,
- `sort=False` – nəticəni birləşmə açarına görə sıralamır,
- `suffixes=('_x', '_y')` – eyni adlı sütunlar üçün əlavə ediləcək sonluqlar,
- `copy=True` – nəticədə yeni obyekt yaradılır,
- `indicator=False` – hansı DataFrame-dən gəldiyini göstərən əlavə sütun (`_merge`) əlavə etmək üçün `True` seçilir,
- `validate=None` – uyğunluq yoxlaması üçün (`'one_to_one'`, `'one_to_many'` və s.).

---

* **`df.join(other, on=None, how='left', lsuffix='', rsuffix='', sort=False)`** – `df` adlı DataFrame-ə `other` adlı DataFrame və ya Series-i indeksə və ya müəyyən açar sütuna əsasən birləşdirir.

- `other` – birləşdiriləcək ikinci DataFrame və ya Series,
- `on=None` – `df`-dəki hansı sütuna əsasən birləşdirmə aparılacağını göstərir; `None` olduqda indeksə görə birləşdirilir,
- `how='left'` – birləşdirmə üsulu: `'left'` (default), `'right'`, `'outer'`, `'inner'`,
- `lsuffix=''` – `df`-də eyni adlı sütunlar varsa, fərqləndirmək üçün bu sonluq əlavə olunur,
- `rsuffix=''` – `other`-də eyni adlı sütunlar varsa, fərqləndirmək üçün bu sonluq əlavə olunur,
- `sort=False` – nəticəni indeksə görə sıralamır; `True` seçilərsə, sıralama aparılır.


> `df.join(other)` – **`other` DataFrame yalnız `index`-inə görə birləşdirilə bilər.** Sütuna görə birləşdirmə üçün ya `set_index()`, ya da `merge()` istifadə edilməlidir.
