## Combining and Merging Datasets, concat_Theory


**`np.concatenate((a1, a2, ...), axis=0, out=None, dtype=None, casting="same_kind")`** – Numpy array-ləri ard-arda birləşdirir.

* `a1, a2, ...` – birləşdiriləcək array-lər (tuple və ya list içində),
* `axis=0` – hansı ox üzrə birləşdiriləcək: `0` – satır, `1` – sütun,
* `out=None` – nəticə üçün çıxış array-i (istəyə bağlı),
* `dtype=None` – nəticənin tipini müəyyən edir,
* `casting="same_kind"` – tip çevrilmə qaydası (`'no'`, `'equiv'`, `'safe'`, `'same_kind'`, `'unsafe'`).

---

* **`pd.concat(objs, axis=0, join='outer', ignore_index=False, keys=None, levels=None, names=None, verify_integrity=False, sort=False, copy=True)`** – Pandas obyektlərini (`Series`, `DataFrame`) təyin olunmuş ox (axis) üzrə birləşdirir.

- `objs` – birləşdiriləcək obyektlərin siyahısı və ya dict-i,
- `axis=0` – birləşmənin aparılacağı ox (`0` – satır üzrə, `1` – sütun üzrə),
- `join='outer'` – birləşdirmə üsulu: `'outer'` (bütün indeksləri saxlayır), `'inner'` (yalnız ortaq indeksləri saxlayır),
- `ignore_index=False` – `True` olarsa, indeks sıfırlanır,
- `keys=None` – nəticədə hər obyekt üçün üst səviyyə etiketi əlavə edir (multiindex üçün),
- `levels=None`, `names=None` – multiindex üçün səviyyə və adların verilməsi,
- `verify_integrity=False` – indekslərin təkrarlanmadığını yoxlayır,
- `sort=False` – `join='outer'` zamanı indeks/sütunları sıraya düzür,
- `copy=True` – obyektlərin kopyasını yaradır.
