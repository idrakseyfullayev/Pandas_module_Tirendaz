## Reshaping and Pivoting, stack, unstack, pivot, melt_Theory

* **`df.stack(level=-1, dropna=True)`** – DataFrame-in sütun səviyyələrini (column level) aşağı səviyyəli indekslərə (row level) çevirir, yəni **"uzun formada"** göstərir.

- `level=-1` – hansı səviyyəni stack edəcəyini göstərir (default: ən sonuncu sütun səviyyəsi),
- `dropna=True` – `NaN` dəyərləri olan sətirləri çıxarır (əgər `False` olsa, saxlayır).

> Nəticə `Series` və ya `DataFrame` (əgər multi-level varsa) şəklində olur.
> Əsasən "wide → long" format çevirmələrində istifadə olunur.

---

* **`df.unstack(level=-1, fill_value=None)`** – DataFrame və ya Series-in sətrlərdəki (row index) bir səviyyəsini sütun səviyyəsinə (column level) çevirir, yəni **"uzun formadan geniş formaya"** çevirmə edir.

- `level=-1` – hansı səviyyənin unstack ediləcəyini göstərir (default: ən son səviyyə),
- `fill_value=None` – boş (`NaN`) yerlərin doldurulması üçün dəyər (əgər verilərsə, `NaN`-lar həmin dəyərlə əvəz olunur).

> Nəticə adətən DataFrame olur.
> `stack()` funksiyasının əksi kimi istifadə olunur.

---

* **`df.pivot(index=None, columns=None, values=None)`** – DataFrame-də verilənlərdən yeni pivot cədvəl yaradır, yəni müəyyən `index` və `columns` üzrə `values` sütununu geniş formada təşkil edir.

- `index` – yeni cədvəldə satırlar üçün istifadə olunacaq sütun adı və ya adlar,
- `columns` – yeni cədvəldə sütunlar üçün istifadə olunacaq sütun adı,
- `values` – pivot cədvəldə doldurulacaq dəyərlər üçün sütun adı (əgər göstərilməzsə, bütün sütunlar istifadə olunur).

> Qeyd: `pivot()` istifadə edərkən `index` və `columns` birlikdə unikal kombinasiya yaratmalıdır, əks halda `ValueError` yaranır.
> Əgər təkrarlanan dəyərlər varsa, `pivot_table()` funksiyasından istifadə etmək məsləhətdir.

---

* **`pd.melt(frame, id_vars=None, value_vars=None, var_name=None, value_name='value', col_level=None, ignore_index=True)`** – Geniş formada olan DataFrame-i **uzun formata** çevirir (yəni sütunları sətirlərə çevirir).

- `frame` – çevriləcək DataFrame,
- `id_vars` – eyni qalacaq sütun(lar) (məsələn, identifikatorlar),
- `value_vars` – əridiləcək sütunlar (əgər göstərilməzsə, `id_vars` istisna olmaqla qalan bütün sütunlar istifadə olunur),
- `var_name` – yeni dəyişən adları sütununa veriləcək ad (default: `variable`),
- `value_name='value'` – dəyərlərin yazılacağı sütunun adı,
- `col_level` – sütunlar multi-leveldirsə, hansı səviyyəni istifadə etmək lazımdır,
- `ignore_index=True` – indeks sıfırlanır (`False` olarsa, mövcud indeks saxlanılır).

> Əks proses üçün `pivot()` və ya `pivot_table()` istifadə olunur.

---


* **`df.pivot_table(data=None, values=None, index=None, columns=None, aggfunc='mean', fill_value=None, margins=False, dropna=True, margins_name='All', observed=False, sort=True)`** – Təkrarlanan dəyərləri **qruplaşdıraraq xülasə** (aggrega­siya) edən **pivot cədvəl** yaradır.

- `data` – istifadə olunacaq DataFrame (əksər hallarda `df.pivot_table(...)` şəklində `data` lazım olmur),
- `values` – hansı sütunun dəyərləri ilə hesablama aparılacaq,
- `index` – satırları təyin edən sütun(lar),
- `columns` – sütunları təyin edən sütun,
- `aggfunc='mean'` – tətbiq olunacaq funksiyalar (məsələn: `'mean'`, `'sum'`, `'count'`, `np.min`, `list` və s.),
- `fill_value=None` – boş (`NaN`) dəyərlərin əvəzlənməsi üçün dəyər,
- `margins=False` – `True` olarsa, satır və sütunların cəmlərini əlavə edir ("All"),
- `dropna=True` – yalnız dolu sütun/satır kombinasiyalarını saxlayır,
- `margins_name='All'` – margin (cəmlər) üçün istifadə olunacaq ad,
- `observed=False` – kateqorik dəyişənlərdə yalnız mövcud olan kombinasiyalar saxlanılsınmı?,
- `sort=True` – nəticəni sıralayır.

> `pivot()` funksiyasından fərqli olaraq `pivot_table()` təkrarlanan açar kombinasiyalarında **qruplama və aqreqasiya** tətbiq edir.
